```python
# %% Cell 1 - Imports and Setup
import ast
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from PIL import Image
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms
from tqdm.auto import tqdm

device = torch.device("cuda:2" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
print(f"PyTorch version: {torch.__version__}")

# %% Cell 2 - Configuration & Paths
path = "/datasets/multi-view-pig-posture-recognition"
train2_path = f"{path}/train2.csv"
train2_images_path = f"{path}/train2_images"
test_path = f"{path}/test.csv"
test_images_path = f"{path}/test_images"
sample_submission_path = f"{path}/sample_submission.csv"
txt_path = f"{path}/pig_posture_classes.txt"

BATCH_SIZE = 32
NUM_EPOCHS = 15
LR = 1e-4
WEIGHT_DECAY = 1e-4
VAL_SPLIT = 0.2
SEED = 42
IMG_SIZE = 224

torch.manual_seed(SEED)

# %% Cell 3 - Explore Data
train2_df = pd.read_csv(train2_path)
test_df = pd.read_csv(test_path)
sample_submission_df = pd.read_csv(sample_submission_path)

print("train2 shape:", train2_df.shape)
print("test shape:", test_df.shape)
print("\ntrain2 columns:", train2_df.columns.tolist())
print("\ntrain2 head:")
print(train2_df.head())

with open(txt_path, "r") as f:
    class_name_list = [line.strip() for line in f.readlines() if line.strip()]
print(f"\nClasses: {class_name_list}")

class_id_to_name = {i: name for i, name in enumerate(class_name_list)}


# %% Cell 4 - Dataset
class PigDataset(Dataset):
    def __init__(self, df, images_path, transform=None):
        self.df = df.reset_index(drop=True)
        self.images_path = images_path
        self.transform = transform
        if len(self.df) > 0 and isinstance(self.df["bbox"].iloc[0], str):
            self.df = self.df.copy()
            self.df["bbox"] = self.df["bbox"].apply(ast.literal_eval)
        self.classes = sorted(self.df["class_id"].unique().tolist())
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}
        self.num_classes = len(self.classes)
        print(
            f"Dataset: {len(self.df)} samples, {self.num_classes} classes: {self.classes}"
        )

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = Path(self.images_path) / str(row["image_id"])
        image = Image.open(img_path).convert("RGB")
        bbox = row["bbox"]
        cx, cy, w, h = float(bbox[0]), float(bbox[1]), float(bbox[2]), float(bbox[3])
        x1 = max(0, cx - w / 2)
        y1 = max(0, cy - h / 2)
        x2 = min(image.width, cx + w / 2)
        y2 = min(image.height, cy + h / 2)
        image = image.crop((x1, y1, x2, y2))
        if self.transform:
            image = self.transform(image)
        label = self.class_to_idx[row["class_id"]]
        return image, label


# %% Cell 5 - Transforms & DataLoaders
imagenet_normalize = transforms.Normalize(
    mean=[0.485, 0.456, 0.406],
    std=[0.229, 0.224, 0.225],
)

train_transform = transforms.Compose(
    [
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(15),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
        transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),
        transforms.ToTensor(),
        imagenet_normalize,
    ]
)

val_transform = transforms.Compose(
    [
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        imagenet_normalize,
    ]
)

# Train/val split on train2 only
indices = torch.randperm(
    len(train2_df), generator=torch.Generator().manual_seed(SEED)
).tolist()
train_size = len(train2_df) - int(len(train2_df) * VAL_SPLIT)

train_dataset = PigDataset(
    train2_df.iloc[indices[:train_size]], train2_images_path, transform=train_transform
)
val_dataset = PigDataset(
    train2_df.iloc[indices[train_size:]], train2_images_path, transform=val_transform
)

num_classes = train_dataset.num_classes
class_names = train_dataset.classes
print(f"Classes ({num_classes}): {class_name_list}")
print(f"Train: {len(train_dataset)}, Val: {len(val_dataset)}")

img, lbl = train_dataset[0]
print(f"Sample shape: {img.shape}, label: {lbl} ({class_name_list[lbl]})")

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=4, pin_memory=True
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=4, pin_memory=True
)

# %% Cell 6 - ConvNeXt-Base Model
print("Loading ConvNeXt-Base pretrained on ImageNet...")
model = models.convnext_base(pretrained=True)

in_features = model.classifier[2].in_features
model.classifier[2] = nn.Linear(in_features, num_classes)

for name, param in model.named_parameters():
    param.requires_grad = False

for name, param in model.named_parameters():
    if "features.7" in name or "classifier" in name:
        param.requires_grad = True

model = model.to(device)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total_params = sum(p.numel() for p in model.parameters())
print(f"ConvNeXt-Base loaded.")
print(
    f"Trainable: {trainable:,} / {total_params:,} ({100 * trainable / total_params:.2f}%)"
)

print("\nTrainable layers:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"  {name}: {param.numel():,}")

# %% Cell 7 - Training
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=LR,
    weight_decay=WEIGHT_DECAY,
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS)

history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_val_acc = 0.0
MODEL_SAVE_PATH = "convnext_base_pig_posture.pth"

for epoch in range(NUM_EPOCHS):
    # --- Train ---
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for images, labels in tqdm(
        train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Train]", leave=False
    ):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)
        correct += (outputs.argmax(1) == labels).sum().item()
        total += labels.size(0)
    train_loss = running_loss / total
    train_acc = correct / total
    scheduler.step()

    # --- Val ---
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for images, labels in tqdm(
            val_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Val]", leave=False
        ):
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * images.size(0)
            correct += (outputs.argmax(1) == labels).sum().item()
            total += labels.size(0)
    val_loss = running_loss / total
    val_acc = correct / total

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        print(f"  ✅ New best model saved! Val Acc: {val_acc:.4f}")

    print(
        f"Epoch {epoch+1}/{NUM_EPOCHS} | "
        f"Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} | "
        f"LR: {scheduler.get_last_lr()[0]:.6f}"
    )

print(f"\n🏆 Best Val Acc: {best_val_acc:.4f}")

# %% Cell 8 - Plot Training History
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history["train_loss"], label="Train", linewidth=2)
ax1.plot(history["val_loss"], label="Val", linewidth=2)
ax1.set_title("Loss (ConvNeXt-Base, train2 only)", fontsize=14)
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.legend()
ax1.grid(True)

ax2.plot(history["train_acc"], label="Train", linewidth=2)
ax2.plot(history["val_acc"], label="Val", linewidth=2)
ax2.set_title("Accuracy (ConvNeXt-Base, train2 only)", fontsize=14)
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Accuracy")
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

print(
    f"\nBest Val Acc: {best_val_acc:.4f} at epoch {np.argmax(history['val_acc']) + 1}"
)


# %% Cell 9 - Test Inference & Submission
class PigPostureTestDataset(Dataset):
    def __init__(self, dataframe, images_path, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.images_path = images_path
        self.transform = transform
        if len(self.df) > 0 and isinstance(self.df["bbox"].iloc[0], str):
            self.df = self.df.copy()
            self.df["bbox"] = self.df["bbox"].apply(ast.literal_eval)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = Path(self.images_path) / str(row["image_id"])
        image = Image.open(img_path).convert("RGB")

        bbox = row["bbox"]
        cx, cy, w, h = float(bbox[0]), float(bbox[1]), float(bbox[2]), float(bbox[3])
        x1 = max(0, cx - w / 2)
        y1 = max(0, cy - h / 2)
        x2 = min(image.width, cx + w / 2)
        y2 = min(image.height, cy + h / 2)
        image = image.crop((x1, y1, x2, y2))

        if self.transform:
            image = self.transform(image)
        return image, str(row["row_id"])


test_dataset = PigPostureTestDataset(test_df, test_images_path, val_transform)
test_loader = DataLoader(
    test_dataset, batch_size=64, shuffle=False, num_workers=4, pin_memory=True
)

print("Loading best ConvNeXt-Base model...")
model.load_state_dict(torch.load(MODEL_SAVE_PATH, map_location=device))
model.eval()
print("Model loaded successfully!")

print("Running inference on test data...")
predictions = []
row_ids = []

with torch.no_grad():
    for inputs, batch_row_ids in tqdm(test_loader, desc="Predicting"):
        inputs = inputs.to(device)
        outputs = model(inputs)
        _, predicted = torch.max(outputs, 1)
        predictions.extend(predicted.cpu().numpy().tolist())
        row_ids.extend(batch_row_ids)

submission_df = pd.DataFrame(
    {
        "row_id": row_ids,
        "class_id": predictions,
    }
)
submission_df = submission_df.sort_values("row_id").reset_index(drop=True)

print(f"\nSubmission Statistics:")
print(f"Total predictions: {len(submission_df)}")
print(f"Unique row_ids: {submission_df['row_id'].nunique()}")

print(f"\nClass distribution:")
class_counts = submission_df["class_id"].value_counts().sort_index()
for class_id, count in class_counts.items():
    name = class_id_to_name.get(class_id, f"Unknown_{class_id}")
    print(
        f"  {name} (ID {class_id}): {count} samples ({100 * count / len(submission_df):.1f}%)"
    )

print(f"\nFirst 10 predictions:")
print(submission_df.head(10))

sample_row_ids = set(sample_submission_df["row_id"].astype(str))
submission_row_ids = set(submission_df["row_id"].astype(str))
print(f"\nRow IDs match sample submission: {sample_row_ids == submission_row_ids}")
print(f"Expected rows: {len(sample_submission_df)}, Actual rows: {len(submission_df)}")

submission_filename = "submission.csv"
submission_df.to_csv(submission_filename, index=False)
print(f"\n✅ Submission saved as '{submission_filename}'")
```

```python
# language: python
# Cell: Improved training - unfreeze more of ConvNeXt + class-balanced sampler + weighted loss + discriminative LR

from collections import Counter

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from torchvision import models

device = torch.device("cuda:2" if torch.cuda.is_available() else "cpu")
print("Device:", device, "PyTorch:", torch.__version__)

# ASSUMPTION: train_dataset and val_dataset already created as in your notebook (PigDataset instances)
# If not, recreate them as before.

# 1) Compute class counts & weights
all_labels = train_dataset.df["class_id"].map(train_dataset.class_to_idx).tolist()
class_counts = Counter(all_labels)
num_classes = train_dataset.num_classes
print("Class counts:", class_counts)

# inverse frequency weights
class_weights = []
total = float(len(all_labels))
for c in range(num_classes):
    cnt = class_counts.get(c, 0)
    if cnt == 0:
        class_weights.append(0.0)
    else:
        class_weights.append(total / (num_classes * cnt))
class_weights = torch.tensor(class_weights, dtype=torch.float32).to(device)
print("Class weights:", class_weights.cpu().numpy())

# 2) Create sample weights and WeightedRandomSampler for train loader
sample_weights = [class_weights[label].item() for label in all_labels]
sampler = WeightedRandomSampler(
    weights=sample_weights, num_samples=len(sample_weights), replacement=True
)

# DataLoader with sampler (do not shuffle when using sampler)
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=4,
    pin_memory=True,
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=4, pin_memory=True
)

# 3) Load ConvNeXt and unfreeze features.6 + features.7 + classifier
model = models.convnext_base(pretrained=True)
in_features = model.classifier[2].in_features
model.classifier[2] = nn.Linear(in_features, num_classes)

# Freeze all first
for name, param in model.named_parameters():
    param.requires_grad = False

# Unfreeze features.6, features.7, and classifier
for name, param in model.named_parameters():
    if (
        name.startswith("features.6")
        or name.startswith("features.7")
        or name.startswith("classifier")
    ):
        param.requires_grad = True

model = model.to(device)

# 4) Set up loss with class weights and optimizer with discriminative LR
criterion = nn.CrossEntropyLoss(weight=class_weights)

# Parameter groups: backbone (features.6+7) small lr, head (classifier) base lr
backbone_lr = LR * 0.1
head_lr = LR

backbone_params = []
head_params = []
for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if "classifier" in name:
        head_params.append(param)
    else:
        backbone_params.append(param)

optimizer = torch.optim.AdamW(
    [
        {"params": backbone_params, "lr": backbone_lr},
        {"params": head_params, "lr": head_lr},
    ],
    weight_decay=WEIGHT_DECAY,
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS)

# 5) Training loop with gradient clipping
best_val_acc = 0.0
MODEL_SAVE_PATH = "convnext_base_train2_balanced.pth"
history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}

for epoch in range(NUM_EPOCHS):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for images, labels in tqdm(
        train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Train]", leave=False
    ):
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        # gradient clipping
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_acc = correct / total
    scheduler.step()

    # validation
    model.eval()
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for images, labels in tqdm(
            val_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Val]", leave=False
        ):
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_running_loss += loss.item() * images.size(0)
            preds = outputs.argmax(dim=1)
            val_correct += (preds == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_running_loss / val_total
    val_acc = val_correct / val_total

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        print(f"  ✅ New best model saved! Val Acc: {val_acc:.4f}")

    print(
        f"Epoch {epoch+1}/{NUM_EPOCHS} | "
        f"Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} | "
        f"LRs: backbone {backbone_lr:.2e}, head {head_lr:.2e}"
    )

print("Finished. Best val acc:", best_val_acc)

# OPTIONAL: after training, load best model and run inference as before
```

```python
# %% Cell 8 - Plot Training History
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history["train_loss"], label="Train", linewidth=2)
ax1.plot(history["val_loss"], label="Val", linewidth=2)
ax1.set_title("Loss (ConvNeXt-Base, train2 only)", fontsize=14)
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.legend()
ax1.grid(True)

ax2.plot(history["train_acc"], label="Train", linewidth=2)
ax2.plot(history["val_acc"], label="Val", linewidth=2)
ax2.set_title("Accuracy (ConvNeXt-Base, train2 only)", fontsize=14)
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Accuracy")
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

print(
    f"\nBest Val Acc: {best_val_acc:.4f} at epoch {np.argmax(history['val_acc']) + 1}"
)


# %% Cell 9 - Test Inference & Submission
class PigPostureTestDataset(Dataset):
    def __init__(self, dataframe, images_path, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.images_path = images_path
        self.transform = transform
        if len(self.df) > 0 and isinstance(self.df["bbox"].iloc[0], str):
            self.df = self.df.copy()
            self.df["bbox"] = self.df["bbox"].apply(ast.literal_eval)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = Path(self.images_path) / str(row["image_id"])
        image = Image.open(img_path).convert("RGB")

        bbox = row["bbox"]
        cx, cy, w, h = float(bbox[0]), float(bbox[1]), float(bbox[2]), float(bbox[3])
        x1 = max(0, cx - w / 2)
        y1 = max(0, cy - h / 2)
        x2 = min(image.width, cx + w / 2)
        y2 = min(image.height, cy + h / 2)
        image = image.crop((x1, y1, x2, y2))

        if self.transform:
            image = self.transform(image)
        return image, str(row["row_id"])


test_dataset = PigPostureTestDataset(test_df, test_images_path, val_transform)
test_loader = DataLoader(
    test_dataset, batch_size=64, shuffle=False, num_workers=4, pin_memory=True
)

print("Loading best ConvNeXt-Base model...")
model.load_state_dict(torch.load(MODEL_SAVE_PATH, map_location=device))
model.eval()
print("Model loaded successfully!")

print("Running inference on test data...")
predictions = []
row_ids = []

with torch.no_grad():
    for inputs, batch_row_ids in tqdm(test_loader, desc="Predicting"):
        inputs = inputs.to(device)
        outputs = model(inputs)
        _, predicted = torch.max(outputs, 1)
        predictions.extend(predicted.cpu().numpy().tolist())
        row_ids.extend(batch_row_ids)

submission_df = pd.DataFrame(
    {
        "row_id": row_ids,
        "class_id": predictions,
    }
)
submission_df = submission_df.sort_values("row_id").reset_index(drop=True)

print(f"\nSubmission Statistics:")
print(f"Total predictions: {len(submission_df)}")
print(f"Unique row_ids: {submission_df['row_id'].nunique()}")

print(f"\nClass distribution:")
class_counts = submission_df["class_id"].value_counts().sort_index()
for class_id, count in class_counts.items():
    name = class_id_to_name.get(class_id, f"Unknown_{class_id}")
    print(
        f"  {name} (ID {class_id}): {count} samples ({100 * count / len(submission_df):.1f}%)"
    )

print(f"\nFirst 10 predictions:")
print(submission_df.head(10))

sample_row_ids = set(sample_submission_df["row_id"].astype(str))
submission_row_ids = set(submission_df["row_id"].astype(str))
print(f"\nRow IDs match sample submission: {sample_row_ids == submission_row_ids}")
print(f"Expected rows: {len(sample_submission_df)}, Actual rows: {len(submission_df)}")

submission_filename = "submission.csv"
submission_df.to_csv(submission_filename, index=False)
print(f"\n✅ Submission saved as '{submission_filename}'")
```
